#REGRESI BERGANDA - **KOTA MAKASSAR**

###Saya Mengambil Variabel kota Makassar yaitu:

###Variabel dependent (Y) : PAD (Pendapatan asli daerah)
###Variabel independent
###(X1): DBH_PAJAK
###(X2) : DAU_Grant


##IMPORT LIBRARY

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import plotly.graph_objects as go
```

##IMPORT & FILTER KOTA MAKASSAR

```python
#Membaca dan memiih dataset
df = pd.read_csv('dataset_fiskal_97_hipoFull.csv')
# Memfilter khusus untuk Kota Makassar
df_mks = df[df['Kota'] == 'Kota Makassar'].copy()

```

##MENENTUKAN VARIABEL DEPENDEN DAN INDPENDEN

```python
# Variabel Dependen (Y) : PAD (Pendapatan Asli Daerah)
target_y = 'PAD'

# Variabel Independen
x1_col = 'DBH_PAJAK'
x2_col = 'DAU_Grant' # Ditambahkan sebagai X2 agar bisa membentuk bidang 3D

y = df_mks[target_y]
X1 = df_mks[x1_col]
X2 = df_mks[x2_col]

# Membentuk dataframe independen
X = pd.DataFrame({x1_col: X1, x2_col: X2})
X_sm = sm.add_constant(X)
```

##PERMODELAN REGRESI OLS

```python
model = sm.OLS(y, X_sm).fit()
print(model.summary())

# Mengambil parameter persamaan
b0 = model.params['const']
b1 = model.params[x1_col]
b2 = model.params[x2_col]
r_squared = model.rsquared
```

##VISUALISASI 3D

```python
x1_range = np.linspace(X1.min(), X1.max(), 30)
x2_range = np.linspace(X2.min(), X2.max(), 30)
x1_grid, x2_grid = np.meshgrid(x1_range, x2_range)
y_grid = b0 + b1 * x1_grid + b2 * x2_grid
fig = go.Figure()
# A. Bidang Regresi (Surface)
fig.add_trace(go.Surface(
    x=x1_range, y=x2_range, z=y_grid,
    colorscale='Viridis', opacity=0.7, name='Bidang Regresi'
))

# B. Titik Data Aktual Kota Makassar
fig.add_trace(go.Scatter3d(
    x=X1, y=X2, z=y,
    mode='markers+text',
    marker=dict(size=6, color=y, colorscale='Viridis', opacity=0.9),
    text=df_mks['Tahun'].apply(lambda t: f"Tahun {t}"),
    textposition="top center",
    name='Data Aktual'
))

# C. Mengatur layout
title_text = f"REGRESI 3D – KOTA MAKASSAR<br><sup>Ŷ = {b0:.2f} + ({b1:.2f}) DBH_Pajak + ({b2:.2f}) DAU_Grant | R² = {r_squared:.3f}</sup>"
fig.update_layout(
    title=dict(text=title_text, x=0.5, font=dict(size=14)),
    scene=dict(
        xaxis_title="DBH Pajak (X₁) [Rp]",
        yaxis_title="DAU Grant (X₂) [Rp]",
        zaxis_title="PAD (Y) [Rp]"
    ),
    width=900, height=700
)

fig.show()
```
